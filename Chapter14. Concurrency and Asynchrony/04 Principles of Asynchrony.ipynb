<h3><strong> Synchronous، Concurrency و Asynchronous </strong></h3>
<h3><strong>1. Synchronous (همزمان)</strong></h3>
<ul><li><p><strong>تعریف:</strong></p><ul><li>در برنامه‌نویسی همزمان، هر عملیات باید به‌طور کامل انجام شود تا کنترل به عملیات بعدی منتقل شود.</li><li>به عبارت دیگر، تا زمانی که یک کار تکمیل نشده، هیچ کار دیگری نمی‌تواند شروع شود.</li></ul></li><li><p><strong>ویژگی‌ها:</strong></p><ul><li>اجرای ترتیبی.</li><li><strong>Blocking:</strong> نخ جاری در طول عملیات بلاک می‌شود.</li></ul></li><li><p><strong>مثال:</strong>
تصور کنید در یک رستوران، پیشخدمت فقط می‌تواند یک سفارش را بگیرد، منتظر آماده شدن غذا باشد و سپس سفارش را به مشتری برگرداند. تا زمانی که سفارش اول تمام نشود، نمی‌تواند به مشتری دیگری خدمت کند.</p></li></ul>
</div>

```csharp
void SynchronousOperation()
{
    Console.WriteLine("Start");
    Thread.Sleep(3000); // شبیه‌سازی عملیات طولانی
    Console.WriteLine("End");
}
```

<div dir="rtl" style="margin:auto; width:90%; font-family:vazirmatn; text-align:justify">
<h4><strong>توضیح:</strong></h4>
<ul><li>نخ (Thread) جاری در طول <code>Thread.Sleep</code> بلاک می‌شود و هیچ کار دیگری نمی‌تواند انجام دهد.</li></ul>
<hr>
<h3><strong>2. Concurrency (همزمانی)</strong></h3>
<ul><li><p><strong>تعریف:</strong></p><ul><li>همزمانی به معنای توانایی اجرای چند کار به صورت موازی است.</li><li>این کار ممکن است با استفاده از <strong>چند نخ (Threads)</strong> یا مکانیزم‌های دیگر انجام شود.</li></ul></li><li><p><strong>ویژگی‌ها:</strong></p><ul><li><strong>Parallelism ممکن است:</strong> همزمانی می‌تواند منجر به اجرای موازی شود، اما لزوماً همه‌ی عملیات‌ها به صورت دقیقاً همزمان اجرا نمی‌شوند.</li><li>از منابع سیستم به صورت بهینه‌تر استفاده می‌شود.</li></ul></li><li><p><strong>مثال:</strong>
در رستوران، پیشخدمت می‌تواند به چند مشتری به صورت همزمان خدمت کند. او سفارش اول را می‌گیرد، سفارش دوم را می‌گیرد، و سپس به آشپزخانه می‌رود و منتظر آماده شدن هر دو سفارش می‌ماند.</p></li></ul>
<div>

```csharp
void ConcurrentOperations()
{
    Task.Run(() => 
    {
        Thread.Sleep(3000); // شبیه‌سازی کار اول
        Console.WriteLine("Task 1 completed");
    });

    Task.Run(() =>
    {
        Thread.Sleep(2000); // شبیه‌سازی کار دوم
        Console.WriteLine("Task 2 completed");
    });

    Console.WriteLine("Both tasks started");
}
```

<div dir="rtl" style="margin:auto; width:90%; font-family:vazirmatn; text-align:justify">
<h4><strong>توضیح:</strong></h4>
<ul><li>دو Task به صورت همزمان شروع می‌شوند و هرکدام روی یک نخ جداگانه اجرا می‌شوند.</li><li>ترتیب اجرای "Task 1 completed" و "Task 2 completed" ممکن است به مدت زمان اجرای هر Task بستگی داشته باشد.</li></ul>
<hr>
<h3><strong>3. Asynchronous (ناهمزمان)</strong></h3>
<ul><li><p><strong>تعریف:</strong></p><ul><li>در برنامه‌نویسی ناهمزمان، عملیات می‌تواند بدون بلاک کردن نخ جاری اجرا شود.</li><li>عملیات طولانی‌مدت (مانند خواندن فایل یا انتظار برای پاسخ شبکه) بدون استفاده از نخ‌های اضافی مدیریت می‌شوند.</li></ul></li><li><p><strong>ویژگی‌ها:</strong></p><ul><li>نخ جاری آزاد است تا به کارهای دیگر بپردازد.</li><li>با <strong>async/await</strong> یا سایر مکانیزم‌های مبتنی بر رویداد پیاده‌سازی می‌شود.</li></ul></li><li><p><strong>مثال:</strong>
در رستوران، پیشخدمت سفارش مشتری را به آشپزخانه می‌دهد و بلافاصله به سراغ مشتری بعدی می‌رود. وقتی غذا آماده شد، آشپزخانه به او اطلاع می‌دهد و او سفارش را تحویل می‌دهد.</p></li></ul>
</div>

```csharp
async Task AsynchronousOperation()
{
    Console.WriteLine("Start");
    await Task.Delay(3000); // شبیه‌سازی عملیات طولانی
    Console.WriteLine("End");
}
```

<div dir="rtl" style="margin:auto; width:90%; font-family:vazirmatn; text-align:justify">
<h4><strong>توضیح:</strong></h4>
<ul><li>متد <code>Task.Delay</code> بدون بلاک کردن نخ اجرا می‌شود.</li><li>نخ جاری می‌تواند در این زمان به کارهای دیگر بپردازد.</li></ul>
</div>

```csharp
using System;
using System.Net.Http;
using System.Threading;

class Program
{
    static string apiResponse = null; // متغیر اشتراکی برای نگه داشتن پاسخ API
    static ManualResetEvent resetEvent = new ManualResetEvent(false); // برای همگام‌سازی برنامه اصلی و Thread

    static void Main(string[] args)
    {
        Console.WriteLine("Starting program...");

        // ایجاد Thread برای فراخوانی API
        Thread apiThread = new Thread(CallApiSynchronously);
        apiThread.Start();

        // منتظر می‌مانیم تا Thread کار خود را تمام کند
        resetEvent.WaitOne();

        // پاسخ را در برنامه اصلی دریافت و نمایش می‌دهیم
        Console.WriteLine("API Response received in Main Thread:");
        Console.WriteLine(apiResponse);

        Console.WriteLine("Program finished.");
    }

    static void CallApiSynchronously()
    {
        Console.WriteLine("Calling API...");
        using (var httpClient = new HttpClient())
        {
            try
            {
                // فراخوانی API و دریافت پاسخ
                var response = httpClient.GetStringAsync("https://jsonplaceholder.typicode.com/posts/1").Result;
                apiResponse = response; // ذخیره پاسخ در متغیر اشتراکی
            }
            catch (Exception ex)
            {
                apiResponse = "Error: " + ex.Message;
            }
            finally
            {
                // اعلام به برنامه اصلی که Thread تمام شده است
                resetEvent.Set();
            }
        }
    }
}
```

```csharp
public async Task<string> FetchDataAsync(string url)
{
    using (var httpClient = new HttpClient())
    {
        // عملیات دانلود غیرهمزمان
        string data = await httpClient.GetStringAsync(url);
        return data;
    }
}

// فراخوانی تابع غیرهمزمان
var dataTask = FetchDataAsync("https://example.com");
// در اینجا برنامه می‌تواند کارهای دیگری انجام دهد
Console.WriteLine("Fetching data...");
string data = await dataTask;
Console.WriteLine(data);
```

### Asynchronous Programming and Continuations

<div dir="rtl" style="margin:auto; width:90%; font-family:vazirmatn; text-align:justify">
<h3><strong>1. نقش Task‌ها در برنامه‌نویسی غیرهمزمان:</strong></h3>
<ul><li>Task‌ها به دلیل پشتیبانی از <strong>continuations</strong> برای برنامه‌نویسی غیرهمزمان بسیار مناسب هستند. Continuations به معنی ادامه اجرای عملیات پس از اتمام یک عملیات غیرهمزمان است.</li><li>در مثال، متدی به نام <strong>Delay</strong> نوشته شده است که از <strong>TaskCompletionSource</strong> برای پیاده‌سازی عملیات I/O غیرهمزمان استفاده می‌کند.</li></ul>
<h3><strong>2. تفاوت عملیات I/O و Compute-Bound:</strong></h3>
<ul><li>عملیات <strong>I/O-bound</strong>: به عملیاتی گفته می‌شود که بیشتر منتظر منابع خارجی (مثل فایل، شبکه، یا دیتابیس) هستند و کمترین استفاده را از پردازنده دارند. این نوع عملیات اغلب از TaskCompletionSource استفاده می‌کنند.</li><li>عملیات <strong>Compute-bound</strong>: به عملیاتی اشاره دارد که نیاز به پردازش سنگین روی CPU دارند. برای این نوع عملیات از <strong>Task.Run</strong> استفاده می‌شود تا عملیات در یک <strong>thread</strong> جداگانه اجرا شود.</li></ul>
<h3><strong>3. مثال محاسبه اعداد اول:</strong></h3>

<h3><strong>4. بهبود کد با Asynchronous Programming:</strong></h3>

```csharp
async Task DisplayPrimeCountsAsync()
{
    for (int i = 0; i < 10; i++)
    {
        int result = await Task.Run(() => GetPrimesCount(i * 1000000 + 2, 100000));
        Console.WriteLine(result);
        //show on the window in wpf application 
    }
}
```

### `Awaiter`

<div dir="rtl" style="margin:auto; width:90%; font-family:vazirmatn; text-align:justify">
<p>در سی‌شارپ، <strong>Awaiter</strong> یک مفهوم کلیدی در برنامه‌نویسی ناهمزمان (Asynchronous Programming) است که به همراه <strong>async</strong> و <strong>await</strong> استفاده می‌شود. Awaiter مسئول مدیریت انتظار برای تکمیل یک عملیات ناهمزمان و بازگشت نتیجه آن است. در اینجا به طور مفصل به بررسی Awaiter و نقش آن در سی‌شارپ می‌پردازیم:</p>
<h3>۱. <strong>مفهوم Awaiter</strong></h3>
<ul><li><p><code>Awaiter</code> یک واسط (Interface) یا شیء است که امکان انتظار برای تکمیل یک عملیات ناهمزمان را فراهم می‌کند.</p></li><li><p>وقتی از <code>await</code> روی یک تسک (Task) یا هر شیء دیگری استفاده می‌کنید، کامپایلر سی‌شارپ به طور خودکار یک <code>Awaiter</code> ایجاد می‌کند تا عملیات انتظار و ادامه (Continuation) را مدیریت کند.</p></li><li><p>Awaiter دو وظیفه اصلی دارد:</p><ol><li><p>بررسی اینکه آیا عملیات ناهمزمان تکمیل شده است یا خیر.</p></li><li><p>اگر عملیات تکمیل نشده باشد، ادامه کد (Continuation) را به تعویق بیندازد تا پس از تکمیل عملیات اجرا شود.</p></li></ol></li></ul>
<h3>۲. <strong>چگونه Awaiter کار می‌کند؟</strong></h3>
<ul><li><p>وقتی از <code>await</code> استفاده می‌کنید، کامپایلر سی‌شارپ به طور خودکار متدهای زیر را فراخوانی می‌کند:</p><ol><li><p><code>GetAwaiter()</code>: این متد یک شیء Awaiter برمی‌گرداند. این شیء باید واسط <code>INotifyCompletion</code> یا <code>ICriticalNotifyCompletion</code> را پیاده‌سازی کند.</p></li><li><p><code>IsCompleted</code>: این خاصیت (Property) بررسی می‌کند که آیا عملیات ناهمزمان تکمیل شده است یا خیر. اگر عملیات تکمیل شده باشد، ادامه کد بلافاصله اجرا می‌شود.</p></li><li><p><code>OnCompleted(Action)</code>: اگر عملیات تکمیل نشده باشد، این متد یک ادامه (Continuation) را ثبت می‌کند تا پس از تکمیل عملیات اجرا شود.</p></li><li><p><code>GetResult()</code>: این متد نتیجه عملیات ناهمزمان را برمی‌گرداند.</p></li></ol></li></ul>

</div>

```csharp
Console.WriteLine("before await");
await Task.Delay(1000);
Console.WriteLine("after await");
```

```csharp
Console.WriteLine("before await");
var awaiter = Task.Delay(1000).GetAwaiter();
awaiter.OnCompleted(()=> Console.WriteLine("after await")).Wait();//
```

```csharp
Task<int> task = GetPrimesCountAsync(0, 1000000);

var awaiter = task.GetAwaiter(); // دریافت Awaiter
awaiter.OnCompleted(() => // ثبت ادامه (Continuation)
{
    int result = awaiter.GetResult(); // دریافت نتیجه
    Console.WriteLine(result);
});
```

<div dir="rtl" style="margin:auto; width:90%; font-family:vazirmatn; text-align:justify">
<ul><li><p><code>GetAwaiter()</code> یک Awaiter برای تسک ایجاد می‌کند.</p></li><li><p><code>OnCompleted</code> یک ادامه (Continuation) ثبت می‌کند که پس از تکمیل تسک اجرا می‌شود.</p></li><li><p><code>GetResult()</code> نتیجه تسک را برمی‌گرداند.</p></li></ul>

<h3>۴. <strong>استفاده از Awaiter با async/await</strong></h3>
<p>وقتی از <code>async</code> و <code>await</code> استفاده می‌کنید، کامپایلر سی‌شارپ به طور خودکار Awaiter را مدیریت می‌کند. به عنوان مثال:</p>
</div>

```csharp
async Task DisplayPrimeCountsAsync()
{
    for (int i = 0; i < 10; i++)
    {
        int result = await GetPrimesCountAsync(i * 1000000 + 2, 1000000);
        Console.WriteLine(result + " primes between...");
    }
    Console.WriteLine("Done!");
}
```

<div dir="rtl" style="margin:auto; width:90%; font-family:vazirmatn; text-align:justify">
<ul><li><p><strong>await</strong> به طور خودکار <strong>GetAwaiter()</strong> را فراخوانی می‌کند.</p></li><li><p>اگر تسک تکمیل نشده باشد، ادامه کد به تعویق می‌افتد و پس از تکمیل تسک اجرا می‌شود.</p></li></ul>
<h3>۵. <strong>پیاده‌سازی Awaiter برای کلاس‌های سفارشی</strong></h3>
</div>

```csharp
public class MyTask
{
    public MyAwaiter GetAwaiter()
    {
        return new MyAwaiter();
    }
}

public class MyAwaiter : INotifyCompletion
{
    public bool IsCompleted { get; private set; }

    public void OnCompleted(Action continuation)
    {
        // ثبت ادامه (Continuation)
        Task.Run(() =>
        {
            Thread.Sleep(1000); // شبیه‌سازی یک عملیات ناهمزمان
            IsCompleted = true;
            continuation();
        });
    }

    public int GetResult()
    {
        return 42; // نتیجه دلخواه
    }
}
```

```csharp
async Task UseMyTaskAsync()
{
    MyTask task = new MyTask();
    int result = await task;
    Console.WriteLine(result);
}
```

<div dir="rtl" style="margin:auto; width:90%; font-family:vazirmatn; text-align:justify">
<p>در سی‌شارپ، <code>await</code> روی هر شیئی که یک <code>Awaiter</code> را ارائه دهد، می‌تواند استفاده شود. برای اینکه یک شیء بتواند با <code>await</code> استفاده شود، باید شرایط زیر را داشته باشد:</p>
</div>

### **State Machine** 

<div dir="rtl" style="margin:auto; width:90%; font-family:vazirmatn; text-align:justify">
<p>در سی شارپ، <code>State Machine</code> (ماشین حالت) یک مکانیزم داخلی است که توسط کامپایلر برای مدیریت کدهای <strong>ناهمزمان (async/await)</strong> ایجاد میشود. این State Machine به کامپایلر اجازه میدهد تا اجرای کدهای ناهمزمان را بهصورت ساختاریافته و کارآمد مدیریت کند. در اینجا بهطور کامل توضیح میدهیم که State Machine چیست، چگونه کار میکند، و در چه مواردی استفاده میشود.</p>

<h3>۱. <strong>State Machine چیست؟</strong></h3>
<p>State Machine یک ساختار برنامهنویسی است که وضعیت (state) فعلی یک سیستم را ردیابی میکند و بر اساس رویدادها (events) یا شرایط خاص، از یک حالت به حالت دیگر منتقل میشود. در سی شارپ، State Machine بهطور خودکار توسط کامپایلر برای مدیریت کدهای <code>async/await</code> ایجاد میشود.</p>
<h3>۲. <strong>چرا State Machine در <code>async/await</code> استفاده میشود؟</strong></h3>
<p>هنگامی که از <code>async/await</code> استفاده میکنید، کامپایلر باید بتواند:</p>
<ul><li><p>اجرای کد را در نقطهٔ <code>await</code> متوقف کند.</p></li><li><p>پس از اتمام <code>Task</code>، اجرای کد را از همان نقطه ادامه دهد.</p></li><li><p>وضعیت متغیرهای محلی و جریان اجرا را بین توقف و ادامه حفظ کند.</p></li></ul>
<p>این کار بدون State Machine بسیار پیچیده و دستی خواهد بود. State Machine این فرآیند را بهصورت خودکار و بهینه مدیریت میکند.</p>
<h3>۳. <strong>State Machine چگونه کار میکند؟</strong></h3>
<p>وقتی یک متد <code>async</code> تعریف میکنید، کامپایلر سی شارپ این متد را به یک State Machine تبدیل میکند. این State Machine شامل مراحل زیر است:</p>
<h4>۱. <strong>تقسیم کد به بخشهای مختلف:</strong></h4>
<p>کامپایلر کد متد <code>async</code> را به بخشهای مختلف تقسیم میکند. هر بخش مربوط به یک <code>await</code> است. بهعنوان مثال:</p>
</div>

```csharp
async Task MyMethodAsync()
{
    Console.WriteLine("Start");
    await Task.Delay(1000); // بخش ۱
    Console.WriteLine("After first await");
    await Task.Delay(1000); // بخش ۲
    Console.WriteLine("After second await");
}
```

<div dir="rtl" style="margin:auto; width:90%; font-family:vazirmatn; text-align:justify">
<p>این کد به سه بخش تقسیم میشود:</p>
<ol start="1"><li><p>کد قبل از اولین <code>await</code>.</p></li><li><p>کد بین اولین و دومین <code>await</code>.</p></li><li><p>کد پس از دومین <code>await</code>.</p></li></ol>
<h4>۲. <strong>مدیریت وضعیت (State):</strong></h4>
<p>State Machine یک عدد (state) را ردیابی میکند که نشان میدهد کدام بخش از کد باید اجرا شود. مثلاً:</p>
<ul><li><p><code>state = 0</code>: اجرای بخش اول.</p></li><li><p><code>state = 1</code>: اجرای بخش دوم.</p></li><li><p><code>state = 2</code>: اجرای بخش سوم.</p></li></ul>
<h4>۳. <strong>ادامهٔ اجرا (Continuation):</strong></h4>
<p>وقتی یک <code>Task</code> کامل میشود، State Machine بهطور خودکار اجرای کد را از نقطهای که متوقف شده بود ادامه میدهد. این کار با استفاده از <code>MoveNext</code> انجام میشود.</p>
<h4>۴. <strong>ذخیرهٔ وضعیت متغیرها:</strong></h4>
<p>State Machine وضعیت متغیرهای محلی را بین توقف و ادامهٔ اجرا حفظ میکند. این کار با ذخیرهٔ متغیرها در فیلدهای State Machine انجام میشود.</p>
</div>

```csharp
async Task MyMethodAsync()
{
    int x = 10;
    Console.WriteLine("Start");
    await Task.Delay(1000);
    Console.WriteLine($"After first await: {x}");
    x = 20;
    await Task.Delay(1000);
    Console.WriteLine($"After second await: {x}");
}
```

```csharp
class MyMethodAsyncStateMachine
{
    private int _state = 0;
    private int _x;
    private TaskAwaiter _awaiter;

    public void MoveNext()
    {
        switch (_state)
        {
            case 0:
                Console.WriteLine("Start");
                _awaiter = Task.Delay(1000).GetAwaiter();
                if (!_awaiter.IsCompleted)
                {
                    _state = 1;
                    _awaiter.OnCompleted(MoveNext);
                    return;//
                }
                goto case 1;
            case 1:
                _awaiter.GetResult();
                Console.WriteLine($"After first await: {_x}");
                _x = 20;
                _awaiter = Task.Delay(1000).GetAwaiter();
                if (!_awaiter.IsCompleted)
                {
                    _state = 2;
                    _awaiter.OnCompleted(MoveNext);
                    return;
                }
                goto case 2;
            case 2:
                _awaiter.GetResult();
                Console.WriteLine($"After second await: {_x}");
                break;
        }
    }
}
```

### Why Language Support Is Important

<div dir="rtl" style="margin:auto; width:90%; font-family:vazirmatn; text-align:justify">
<h3>۱. <strong>اهمیت پشتیبانی زبانی (Language Support)</strong></h3>
<ul><li><p>برنامه‌نویسی ناهمزمان بدون استفاده از <strong>async</strong> و <strong>await</strong> می‌تواند بسیار پیچیده و مستعد خطا باشد.</p></li><li><p>در این مثال، هدف این است که متد <strong>DisplayPrimeCounts</strong> به‌گونه‌ای تغییر کند که از نسخه ناهمزمان <strong>GetPrimesCountAsync</strong> استفاده کند.</p></li><li><p>اگر از <strong>async</strong> و <strong>await</strong> استفاده نکنیم، پیاده‌سازی این تغییرات بسیار دشوار و پرخطا خواهد بود.</p></li></ul>
<h3>۲. <strong>مشکل اجرای موازی ناخواسته</strong></h3>
<p>اگر بخواهیم بدون استفاده از <strong>await</strong>، متد <strong>GetPrimesCountAsync</strong> را فراخوانی کنیم، ممکن است با مشکل اجرای موازی ناخواسته مواجه شویم. به عنوان مثال:</p>
</div>

```csharp
for (int i = 0; i < 10; i++)
{
    var awaiter = GetPrimesCountAsync(i * 1000000 + 2, 1000000).GetAwaiter();
    awaiter.OnCompleted(() =>
        Console.WriteLine(awaiter.GetResult() + " primes between... "));
}
Console.WriteLine("Done");
```

<div dir="rtl" style="margin:auto; width:90%; font-family:vazirmatn; text-align:justify">
<p>در این حالت، حلقه به سرعت تکرار می‌شود و تمام ۱۰ عملیات به صورت موازی اجرا می‌شوند. این رفتار نامطلوب است، زیرا:</p>
<ul><li><p>عملیات‌های داخلی <strong>GetPrimesCountAsync</strong> خودشان از موازی‌سازی استفاده می‌کنند.</p></li><li><p>اجرای موازی این عملیات‌ها باعث می‌شود زمان بیشتری برای دیدن اولین نتایج صرف شود و ترتیب نتایج نیز به‌هم بریزد.</p></li></ul>
<h3>۳. <strong>نیاز به اجرای متوالی (Sequential Execution)</strong></h3>
<p>در بسیاری از موارد، نیاز داریم که تسک‌ها به صورت متوالی اجرا شوند. به عنوان مثال، ممکن است <strong>تسک B</strong> به نتیجه <strong>تسک A</strong> وابسته باشد (مانند یک درخواست شبکه که ابتدا نیاز به DNS lookup دارد).</p>
<p>برای اجرای متوالی تسک‌ها، باید حلقه را حذف کنیم و از یک <strong>فراخوانی بازگشتی (Recursive Call)</strong> در ادامه (Continuation) استفاده کنیم:</p>
</div>

```csharp
void DisplayPrimeCounts()
{
    DisplayPrimeCountsFrom(0);
}

void DisplayPrimeCountsFrom(int i)
{
    var awaiter = GetPrimesCountAsync(i * 1000000 + 2, 1000000).GetAwaiter();
    awaiter.OnCompleted(() =>
    {
        Console.WriteLine(awaiter.GetResult() + " primes between...");
        if (++i < 10) DisplayPrimeCountsFrom(i);
        else Console.WriteLine("Done");
    });
}
```

<div dir="rtl" style="margin:auto; width:90%; font-family:vazirmatn; text-align:justify">
<p>با استفاده از <strong>async</strong> و <strong>await</strong>، تمام این پیچیدگی‌ها از بین می‌رود. کد فوق را می‌توان به سادگی زیر نوشت:</p>
</div>

```csharp
async Task DisplayPrimeCountsAsync()
{
    for (int i = 0; i < 10; i++)
        Console.WriteLine(await GetPrimesCountAsync(i * 1000000 + 2, 1000000) +
            " primes between " + (i * 1000000) + " and " + ((i + 1) * 1000000 - 1));
    Console.WriteLine("Done!");
}
```
