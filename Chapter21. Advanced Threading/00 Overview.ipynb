<p>فصل ۲۱ کتاب <strong>C# in a Nutshell</strong> با عنوان <strong>Advanced Threading</strong> به موضوعات پیشرفته‌تر در زمینه‌ی threading در زبان C# می‌پردازد. این فصل ادامه‌ی مباحثی است که در فصل ۱۴ به‌صورت مقدماتی مطرح شده‌اند. در فصل ۱۴، مفاهیم پایه‌ای مانند شروع و پیکربندی یک thread، thread pooling، blocking، spinning و synchronization contexts معرفی شدند. همچنین مفاهیم قفل‌گذاری (locking) و ایمنی در threading (thread safety) و ساختارهای ساده‌ی سیگنال‌دهی مانند <code>ManualResetEvent</code> بررسی شدند.</p><p>در فصل ۲۱، این مفاهیم با جزئیات بیشتری بررسی می‌شوند و موضوعات پیشرفته‌تری مطرح می‌شوند. در این فصل به موارد زیر پرداخته می‌شود:</p><ol start="1"><li><p><strong>Nonexclusive Locking</strong>:</p><ul><li><p>استفاده از <code>Semaphore</code> و قفل‌های خواننده/نویسنده (reader/writer locks) برای مدیریت دسترسی به منابع به‌صورت غیرانحصاری.</p></li></ul></li><li><p><strong>Signaling Constructs</strong>:</p><ul><li><p>بررسی تمام ساختارهای سیگنال‌دهی مانند <code>AutoResetEvent</code>، <code>ManualResetEvent</code>، <code>CountdownEvent</code> و <code>Barrier</code>.</p></li></ul></li><li><p><strong>Lazy Initialization</strong>:</p><ul><li><p>استفاده از <code>Lazy&lt;T&gt;</code> و <code>LazyInitializer</code> برای مقداردهی اولیه‌ی تنبل (lazy initialization) که بهینه‌سازی در مصرف منابع را به همراه دارد.</p></li></ul></li><li><p><strong>Thread-Local Storage</strong>:</p><ul><li><p>استفاده از <code>ThreadStaticAttribute</code>، <code>ThreadLocal&lt;T&gt;</code> و متدهای <code>GetData</code>/<code>SetData</code> برای ذخیره‌سازی داده‌هایی که مختص هر thread هستند.</p></li></ul></li><li><p><strong>Timers</strong>:</p><ul><li><p>بررسی تایمرها برای اجرای عملیات در بازه‌های زمانی مشخص.</p></li></ul></li></ol><p>علاوه بر این، نویسنده اشاره می‌کند که threading موضوع بسیار گسترده‌ای است و برخی مباحث پیشرفته‌تر و تخصصی‌تر به‌صورت آنلاین در دسترس هستند. این مباحث شامل موارد زیر می‌شوند:</p><ul><li><p><strong>Monitor.Wait و Monitor.Pulse</strong>:</p><ul><li><p>برای سناریوهای خاص سیگنال‌دهی.</p></li></ul></li><li><p><strong>Nonblocking Synchronization Techniques</strong>:</p><ul><li><p>تکنیک‌های همگام‌سازی غیرمسدودکننده برای بهینه‌سازی‌های دقیق (مانند <code>Interlocked</code>، memory barriers و <code>volatile</code>).</p></li></ul></li><li><p><strong>SpinLock و SpinWait</strong>:</p><ul><li><p>برای سناریوهای با همزمانی بالا (high-concurrency).</p></li></ul></li></ul><p>این فصل به شما کمک می‌کند تا درک عمیق‌تری از threading در C# پیدا کنید و بتوانید از ابزارها و تکنیک‌های پیشرفته‌تر برای مدیریت موثرتر threadها و همگام‌سازی بین آن‌ها استفاده کنید. اگر به دنبال اطلاعات بیشتر هستید، می‌توانید به لینک ارائه‌شده در کتاب مراجعه کنید.</p></div>
